```python
import numpy as np
print("NumPy version:", np.__version__)
```

```python
# Creating arrays
a = np.array([10, 20, 30, 40, 50])
print("1D array:", a, "shape:", a.shape, "dtype:", a.dtype)

m = np.arange(1, 10).reshape(3, 3)
print("3x3 matrix:")
print(m)
```

```python
# Vectorised operations and simple statistics
scores = np.array([12, 18, 22, 28, 31, 36, 45])
print("Mean:", round(scores.mean(), 2))
print("Std:", round(scores.std(), 2))
print("Min / Max:", scores.min(), scores.max())
print("Scores + 5:", scores + 5)
```
