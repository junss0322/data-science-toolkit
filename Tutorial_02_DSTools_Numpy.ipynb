```python
# WIA1007 Tutorial 02 - Zhang JunSheng (25007575)
import numpy as np
print("NumPy version:", np.__version__)
```

```python
# 1. A numpy array of size 2 x 2
a = np.array([[1, 2], [3, 4]])
print(a)
print("shape:", a.shape)
```

```python
# 2. A random numpy array of size 3 x 4
r = np.random.rand(3, 4)
print(r)
print("shape:", r.shape)
```

```python
# 3. A numpy array of ones and a numpy array of zeroes
ones = np.ones((3, 3))
zeros = np.zeros((3, 3))
print("ones:")
print(ones)
print("zeros:")
print(zeros)
```

```python
# 4. A random numpy array of size 6 x 3, then transpose it
b = np.random.rand(6, 3)
b_t = b.T
print("original shape:", b.shape)
print(b)
print("transposed shape:", b_t.shape)
print(b_t)
```

```python
# 5. Two random arrays of size n x p and p x m, then their dot product
n, p, m = 2, 3, 4
A = np.random.rand(n, p)
B = np.random.rand(p, m)
C = np.dot(A, B)
print("A shape:", A.shape)
print("B shape:", B.shape)
print("A . B shape:", C.shape)
print(C)
```
